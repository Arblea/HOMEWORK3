Question

Explain the difference between:
```python
A * B
and:
A @ B
```

````python
#A * B performs element‑wise multiplication. This means each element in A is multiplied by the element in the same position in B. The shapes of A and B must match.

#A @ B performs matrix multiplication. This uses row‑by‑column dot products following linear algebra rules. The shapes must be compatible: (m × n) @ (n × p) → (m × p).

#They are not interchangeable because A * B does element‑wise operations, while A @ B does true matrix multiplication.
````

Part 9 — Automatic Differentiation

PyTorch provides automatic differentiation using autograd.

````python
x = torch.tensor([[1., 2.],
                  [3., 4.]], requires_grad=True)

# TODO: Calculate y = 3x² + 2
y = 3 * x**2 + 2
print(y)

````

````python
import torch

# Create x with gradients
x = torch.tensor([[1., 2.],
                  [3., 4.]], requires_grad=True)

# Calculate y = 3x² + 2
y = 3 * x**2 + 2

# Sum y
z = y.sum()

# Backprop
z.backward()

# Print results
print(y)
print(z)
print(x.grad)
````

Questions
* What does requires_grad=True mean?
* What does backward() do?
* What is stored in x.grad?
* Why is automatic differentiation important for neural networks?

````python
# Questions and Answers (Autograd)

# What does requires_grad=True mean?
# → It tells PyTorch to track operations on the tensor so it can compute gradients.

# What does backward() do?
# → It computes gradients by following the computation graph and fills .grad.

# What is stored in x.grad?
# → The gradient of the final scalar output (z) with respect to x.

# Why is automatic differentiation important for neural networks?
# → Neural networks need gradients to update weights during training.
#   Autograd computes these gradients automatically and efficiently.
````
