## Part 2 — Indexing and Slicing

Using the array from Part 1:

## Part 3 — Boolean Indexing

Create:
```python
x = np.array([[12, 5, 18, 3],
              [25, 7, 14, 9],
              [2, 30, 11, 20]])
```
1.	Extract all values greater than 15.
2.	Extract all values less than or equal to 10.
3.	Replace every value greater than 20 with 0.
4.	Count how many values are greater than 15.

````python
import numpy as np

x = np.array([[12, 5, 18, 3],
              [25, 7, 14, 9],
              [2, 30, 11, 20]])

# 1. Extract all values greater than 15
print("1.", x[x > 15])

# 2. Extract all values less than or equal to 10
print("2.", x[x <= 10])

# 3. Replace every value greater than 20 with 0
x_mod = x.copy()
x_mod[x_mod > 20] = 0
print("3.", x_mod)

# 4. Count how many values are greater than 15
print("4.", (x > 15).sum())
````

## Part 5 — Matrix Multiplication
Create:
```python
A = np.array([[1, 2, 3],
              [4, 5, 6]])

B = np.array([[7, 8],
              [9, 10],
              [11, 12]])
```
1.	Compute A @ B.
2.	Compute the same result using np.dot(A, B).
3.	Verify that both results are identical.
4.	Print the shape of A, B, and the result.

The notebook demonstrates the equivalence of .dot(), np.dot(), and @ for matrix multiplication.

````python
import numpy as np

A = np.array([[1, 2, 3],
              [4, 5, 6]])

B = np.array([[7, 8],
              [9, 10],
              [11, 12]])

# 1. Compute A @ B
print("1. A @ B")
result1 = A @ B
print(result1)

# 2. Compute the same result using np.dot(A, B)
print("2. np.dot(A, B)")
result2 = np.dot(A, B)
print(result2)

# 3. Verify that both results are identical
print("3. Results identical?")
print(np.array_equal(result1, result2))

# 4. Print the shape of A, B, and the result
print("4. Shapes:")
print("A shape:", A.shape)
print("B shape:", B.shape)
print("Result shape:", result1.shape)
````

## Part 7 — Mini Application: Student Grades

Create a NumPy array containing the grades of 5 students in 4 assignments:

* Student 1: 85, 90, 78, 92
* Student 2: 70, 88, 91, 76
* Student 3: 95, 93, 89, 97
* Student 4: 60, 72, 68, 75
* Student 5: 88, 84, 90, 86

Using NumPy only:
1.	Calculate each student’s average.
2.	Find the highest grade on each assignment.
3.	Find the average grade for each assignment.
4.	Find the student with the highest overall average.
5.	Find all grades below 70.
6.	Replace grades below 70 with 70.
7.	Calculate the overall class average.

````python
import numpy as np

grades = np.array([
    [85, 90, 78, 92],
    [70, 88, 91, 76],
    [95, 93, 89, 97],
    [60, 72, 68, 75],
    [88, 84, 90, 86]
])

# 1. Each student's average
print("1.", np.mean(grades, axis=1))

# 2. Highest grade on each assignment
print("2.", np.max(grades, axis=0))

# 3. Average grade for each assignment
print("3.", np.mean(grades, axis=0))

# 4. Student with highest overall average
student_avgs = np.mean(grades, axis=1)
print("4.", np.argmax(student_avgs) + 1)

# 5. All grades below 70
print("5.", grades[grades < 70])

# 6. Replace grades below 70 with 70
fixed_grades = grades.copy()
fixed_grades[fixed_grades < 70] = 70
print("6.", fixed_grades)

# 7. Overall class average
print("7.", np.mean(fixed_grades))
````

## Challenge Problem

Create a random 5 × 5 matrix containing integers from 1 to 100.

Using NumPy only:
1.	Find the largest value.
2.	Find the smallest value.
3.	Calculate the mean.
4.	Calculate the standard deviation.
5.	Find all values greater than the mean.
6.	Replace all values below the mean with 0.
7.	Calculate the sum of each row and each column.

````python
import numpy as np

# Create random 5x5 matrix
M = np.random.randint(1, 101, (5, 5))
print("Matrix:\n", M)

# 1. Largest value
print("1.", np.max(M))

# 2. Smallest value
print("2.", np.min(M))

# 3. Mean
print("3.", np.mean(M))

# 4. Standard deviation
print("4.", np.std(M))

# 5. Values greater than the mean
mean_val = np.mean(M)
print("5.", M[M > mean_val])

# 6. Replace values below the mean with 0
M_fixed = M.copy()
M_fixed[M_fixed < mean_val] = 0
print("6.\n", M_fixed)

# 7. Sum of each row and each column
print("7. Row sums:", np.sum(M_fixed, axis=1))
print("   Column sums:", np.sum(M_fixed, axis=0))
````
